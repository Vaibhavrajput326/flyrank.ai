## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*
Why ML Beats a Fixed Rule
A fixed rule might look like:

"If traffic drops by more than 20%, refresh the page."

However, page decline is influenced by many factors:

Traffic trends
Engagement metrics
Content characteristics
Client-specific behavior
Machine learning can learn complex relationships between these features.

A fixed rule uses only one or two variables, while ML can combine many signals to make better predictions and prioritize pages more accurately.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
print(
    df["trend_direction"]
    .value_counts()
)

print(
    "Declining Percentage:",
    (
        (df["trend_direction"] == "down")
        .mean()
        * 100
    )
)

print(
    "Unique Clients:",
    df["client_id"].nunique()
)
```
